```python
from abc import ABC, abstractmethod

# ──────────────────────────────────────────────────────────────
#                    ABSTRACT BASE CLASS
# ──────────────────────────────────────────────────────────────

class bank(ABC):
    def __init__(self,name,age1,age2,gender):
        if (name[0].isupper() and name[1]=="." and name[2].isupper() and len(name)<=20):
            self.name=name
        else:
            raise ValueError 
    
        if age1==int(age1) and age1>=12 and age1<=60:
            self.age1=age1
        else:
            raise ValueError 
        
        if gender=="male" or gender=="female":
            self.gender=gender
        else:
            raise ValueError

        if age2.islower() and age2 in (
            "twelve", "thirteen", "fourteen", "fifteen", "sixteen",
            "seventeen", "eighteen", "nineteen", "twenty", "twenty-one",
            "twenty-two", "twenty-three", "twenty-four", "twenty-five",
            "twenty-six", "twenty-seven", "twenty-eight", "twenty-nine",
            "thirty", "thirty-one", "thirty-two", "thirty-three",
            "thirty-four", "thirty-five", "thirty-six", "thirty-seven",
            "thirty-eight", "thirty-nine", "forty", "forty-one",
            "forty-two", "forty-three", "forty-four", "forty-five",
            "forty-six", "forty-seven", "forty-eight", "forty-nine",
            "fifty", "fifty-one", "fifty-two", "fifty-three",
            "fifty-four", "fifty-five", "fifty-six", "fifty-seven",
            "fifty-eight", "fifty-nine", "sixty"
        ):
            self.age2=age2
        else:
            raise ValueError

    @abstractmethod
    def saving_account(self):
        pass

    @abstractmethod
    def joint_account(self):
        pass

    @abstractmethod
    def salary_account(self):
        pass

    @abstractmethod
    def current_account(self):
        pass

    @abstractmethod
    def demat_account(self):
        pass

    def bankname(self):
        print("My bank name is TAMILNADU BANK")

    def location(self):
        print("Egmore")


# ──────────────────────────────────────────────────────────────
#                       SAVINGS ACCOUNT
# ──────────────────────────────────────────────────────────────

class savings(bank):
    def __init__(self,name,age1,age2,gender,guardian,g_name,pancard=None,aadhar=None):
        super(). __init__ (name,age1,age2,gender)
        
        if guardian=="yes":
            self.guardian=guardian
        if g_name[0].isupper() and g_name[1]==".":
            self.g_name=g_name
        else:
            raise ValueError 
            
    def get_pancard(self):
            return self._pancard
    def get_aadhar(self):
            return self._aadhar
            
    def set_pancard(self,value):
        if value.isupper():
            self._pancard=value
        else:
            raise ValueError
    def set_aadhar(self,value):
        if len(value)==12 and value.isdigit():
            self._aadhar=value
        else:
            raise ValueError
            
    def saving_account(self):
        if self.guardian=="yes" and self.age1>=21 and self.age1<=25:
            print("you are eligible for saving aacount")
        elif self.guardian=="yes" and self.age1>25:
            print("create another account")
        elif self.guardian=="yes"  and self.age1<20:
            print("Please add the customer to a joint account")
        else:
            raise ValueError

    def joint_account(self):
        pass

    def salary_account(self):
        pass

    def current_account(self):
        pass

    def demat_account(self):
        pass


# ──────────────────────────────────────────────────────────────
#                        JOINT ACCOUNT
# ──────────────────────────────────────────────────────────────

class joint(bank):
    def __init__(self,name,age1,age2,gender,idcard,fathername,schoolname=None,mothername=None):
        super(). __init__(name,age1,age2,gender)
        
        if idcard=="yes":
            self.idcard=idcard
        else:
            raise ValueError
            
        self.fathername=fathername
        
            
    def get_schoolname(self):
        return self._schoolname
    def get_mothername(self):
        return self._mothername
        
    def set_schoolname(self,value):
        if value.title():
            self.schoolname=value
        else:
            raise ValueError
    def set_mothername(self,value):
        if value[0]==self.fathername[0] and value[1]==".":
            self._mothername=value
        else:
            raise ValueError

    def joint_account(self):
        if self.age1<20 and self.idcard=="yes" and self.schoolname!=None:
            print("you are eligibe for joint account")
        else:
            print("you are not eligible for joint account")
    def saving_account(self):
        pass

    def salary_account(self):
        pass

    def current_account(self):
        pass

    def demat_account(self):
        pass


# ──────────────────────────────────────────────────────────────
#                        SALARY ACCOUNT
# ──────────────────────────────────────────────────────────────

class salary(bank):
    def __init__(self,name,age1,age2,gender,companyname=None,companygrade=None,email_id=None,salary=None):
        super().__init__(name,age1,age2,gender)
        
    def get_companyname(self):
        return self._companyname
    def get_companygrade(self):
        return self._companygrade
    def get_email_id(self):
        return self._email_id
    def get_salary(self):
        return self._salary
    def set_companyname(self,value):
        self._companyname=value
    def set_companygrade(self,value):
        if value=="A" or value=="B" or value=="C":
            self._companygrade=value
        else:
            raise ValueError
    def set_email_id(self,value):                       
        if self._companyname in value:
            self._email_id=value
        else:
            raise ValueError
    def set_salary(self,value):
        if value>0:
            self._salary=value
        else:
            raise ValueError

    def salary_account(self):
        if self._salary>50000 and self._companygrade=="A":
            print("you are eligible for loan upto 1core")
        if self._salary<50000 and self._companygrade=="B":
            print("you are eligible for loan upto 10lakhs")
        if self._salary>50000 and self._companygrade=="C":
            print("you are eligible for loan upto 3lakhs")
        if self._salary<50000 and self._companygrade=="C":
             print("you are eligible for loan upto 1lakhs")
        if self._salary<50000 and self._companygrade=="B":
            print("you are eligible for upto 8lakhs")
    def saving_account(self):
        pass

    def joint_account(self):
        pass

    def current_account(self):
        pass
    def demat_account(self):
        pass


# ──────────────────────────────────────────────────────────────
#                        CURRENT ACCOUNT
# ──────────────────────────────────────────────────────────────

class current(bank):
    def __init__(self,name,age1,age2,gender,business_name=None,gstno=None,revenue=None,aadhar=None,
                 father_aadhar=None,mother_aadhar=None,pancard=None,
                 ownhouse=None,asert=None,cibilscore=None,Auditor=None):
        super().__init__(name,age1,age2,gender)
        
    def get_business_name(self):
        return self._business_name
    def get_gstno(self):
        return self._gstno
    def get_revenue(self):
        return self._revenue
    def get_aadhar(self):
        return self._aadhar
    def get_fatheraadhar(self):
        return self._fatheraadhar
    def get_motheraadhar(self):
        return self._motheraadhar
    def get_pancard(self):
        return self._pancard
    def get_ownhouse(self):
        return self._ownhouse
    def get_asert(self):
        return self._asert
    def get_cibilscore(self):
        return self._cibilscore
    def get_auditor(self):
        return self._auditor

    def set_business_name(self,value):
        self._business_name=value
    def set_gstno(self,value):
        if value[0].isupper():
            self._gstno=value
        else:
            raise ValueError
    def set_revenue(self,value):
        self._revenue=value
    def set_aadhar(self,value):
        if len(value)==12:
           self._aadhar=value
        else:
            raise ValueError
    def set_fatheraadhar(self,value):
         if len(value)==12: 
            self._fatheraadhar=value
         else:
            raise ValueError
    def set_motheraadhar(self,value):
         if len(value)==12: 
            self._motheraadhar=value
         else:
            raise ValueError
    def set_pancard(self,value):
        self._pancard=value
    def set_ownhouse(self,value):
        if value=="yes" or value=="no":
            self._ownhouse=value
        else:
            raise ValueError
    def set_asert(self,value):
        if value[-2:]=="cr":
          self._asert=value
        else:
           raise ValueError
    def set_cibilscore(self,value):
        self._cibilscore=value
    def set_auditor(self,value):
        if value=="yes" or value=="no":
           self._auditor=value
        else:
            raise ValueError
    
    def current_account(self):
         if (
             self.name and self.age1 and self.age2 and self.gender and self._business_name and 
         self._gstno and self._revenue and self._aadhar and self._fatheraadhar and self._motheraadhar and 
         self._pancard and self._ownhouse and self._asert and self._cibilscore and self._auditor 
         ):
             print("all the details are correct")
         else:
             print("enter the valid data")
    def saving_account(self):
        pass

    def joint_account(self):
        pass

    def salary_account(self):
        pass

    def demat_account(self):
        pass



# ──────────────────────────────────────────────────────────────
#                         PROJECT TESTING
# ──────────────────────────────────────────────────────────────

# SAVINGS ACCOUNT TEST

print("\nSAVINGS ACCOUNT")
print("----------------")

jee = savings("J.Malavika",22,"twenty-two","female","yes","J.Jagan")
jee.set_pancard("ABCDE1234F")
jee.set_aadhar("123456789012")
jee.saving_account()


# JOINT ACCOUNT TEST
print("\nJOINT ACCOUNT")
print("-------------")

jee = joint("J.Malavika",20,"twenty","female","yes","Jagan")
jee.set_schoolname("cheran school")
jee.set_mothername("J.sampoo")
jee.joint_account()


# SALARY ACCOUNT TEST

print("\nSALARY ACCOUNT")
print("--------------")

jee = salary("J.Divyaa",20,"twenty","female")
jee.set_companyname("quest")
jee.set_companygrade("A")
jee.set_email_id("maluquest.com")
jee.set_salary(56000)
jee.salary_account()


# CURRENT ACCOUNT TEST
print("\nCURRENT ACCOUNT")
print("---------------")

jee = current("J.Divya",25,"twenty-five","female")
jee.set_business_name("ABCTraders")
jee.set_gstno("A123456789")
jee.set_revenue(5000000)
jee.set_aadhar("123456789012")
jee.set_fatheraadhar("123456789013")
jee.set_motheraadhar("123456789014")
jee.set_pancard("ABCDE1234F")
jee.set_ownhouse("yes")
jee.set_asert("10cr")
jee.set_cibilscore(750)
jee.set_auditor("yes")
jee.current_account()
```
